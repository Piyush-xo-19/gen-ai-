Tools 

model ca call tools to perform task such as fetchig data from a database ,seraching the web or running a code .Tools are pair of 

1 a schema icludig the name of the tool, description and or argument defiitinios often as json

2 a func or a coroutine to exectue


```python
from langchain.chat_models import init_chat_model
model=init_chat_model("gemini-2.5-flash",model_provider="google_genai")
response=model.invoke("why parrots are green ")
```

```python
from langchain.tools import tool

@tool
def get_weather(location:str)->str:
    """get weather at a location""" #it is important to write the doc string of what a func is toing because when we bind this func with llm .llm will get to know ki ye fuc kya krta hai
    return f"ots suny in the {location}"

model_with_tool=model.bind_tools([get_weather])

response=model_with_tool .invoke("whats the weather in the new delhi ")
print(response)
for tool_call in response.tool_call:
    print(f"tool:"{tool_call['name']})  
    print(f"arg:"{tool_call["args"]})  
```

**TOOL EXECUTION LOOPS**



```python
#step 1 model genrates tool call
 messages=[{"role":"user","content":"whats the weather in delhi "}]
ai_msg=model_with_tools.invoke(messages)
messages.append(ai_msg)

#step 2 execute tool and collect results 
for tool_call in ai_msg.tool_cals:
    tool_result=get_weather.invoke(tool_call)
    messages.append(tool_result)

#step 3 pass result back to model for final response 
final_response=model_with_tools.invoke(messages)
print(final_response.txt)

```
